```python
p_num = int("".join(filter(str.isdigit, "3w")))
p_unit = "".join(filter(str.isalpha, "3w")).upper()

offset_map = {"D": "days", "W": "weeks", "M": "months"}
unit_key = offset_map.get(p_unit, "days")

end_date = pd.to_datetime("2026-03-08") + pd.DateOffset(**{unit_key: p_num})
print(end_date)
```
